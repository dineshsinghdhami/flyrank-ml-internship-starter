## 1. My lane as an ML task (type)

*Classification, clustering, ranking, or scoring — which one, and why?*

### ML Task Type

I frame this lane primarily as a **ranking task**.

The goal is to rank pages by how strongly they appear to need attention, so that the most promising content-refresh opportunities can be reviewed first.

Instead of only assigning each page to a yes/no class, the model should produce a score that can be used to order pages from highest to lowest priority. This makes the output useful for deciding which pages should be reviewed first when time and resources are limited.

```python
task_type = "ranking"

print("ML task type:", task_type)
print("Decision supported: prioritize pages for content review")
```

## 2. Target or proxy

### Target or Proxy

The target is a **content refresh priority score** for each page.

Because the dataset may not contain a direct label saying whether a page should be refreshed, I would use an observed or derived proxy based on signals such as declining traffic, outdated content indicators, engagement changes, or other available page-level performance features.

The prediction should help estimate how strongly a page needs review, so pages with higher scores can be prioritized for content refresh.

```python
target_name = "content_refresh_priority"

print("Target / proxy:", target_name)
print("Interpretation: higher score = higher priority for content review")
```

## 3. Success metric

*One metric you can defend. What number means 'good'?*

### Success Metric

I would use **NDCG@10 (Normalized Discounted Cumulative Gain at 10)** as the main success metric.

Because this is a ranking task, the most important pages should appear near the top of the ranked list. NDCG@10 measures how well the top 10 predicted pages match the ideal priority order while giving more importance to pages ranked near the top.

A higher NDCG@10 score means the model is doing a better job of placing the most relevant content-refresh opportunities first.

For this task, I would consider an NDCG@10 score above **0.70** a useful initial result, while still comparing it against a simple baseline ranking.

```python
success_metric = "NDCG@10"
initial_good_threshold = 0.70

print("Success metric:", success_metric)
print("Initial useful threshold:", initial_good_threshold)
print("Goal: rank the highest-priority pages near the top of the review queue")
```

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

### Why ML Beats a Fixed Rule

A fixed rule would be too limited because content-refresh priority can depend on several signals at the same time.

For example, a page may have declining performance but still receive strong engagement, while another page may have weaker engagement, older content, or different traffic patterns. A simple rule such as "refresh every page below one traffic threshold" would ignore these interactions.

Machine learning can combine multiple page-level features and learn patterns from the data, making the ranking more flexible and useful for decision support.
